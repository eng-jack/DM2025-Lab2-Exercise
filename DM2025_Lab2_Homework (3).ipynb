## 1. Preprocessing Steps

前處理的主要目的是將課程提供的三份資料整理成可直接使用的訓練與測試集。首先，我將結構較深的 final_posts.json 展開，從中抽取每則貼文的編號與文字內容，轉成格式統一的表格。接著讀入資料切分檔與情緒標籤，透過貼文編號將三份資料合併。

其中，訓練資料會同時包含貼文文字與對應的情緒標籤；測試資料則僅保留貼文 ID 與文字內容，用於送交 Kaggle 預測。完成後即可得到乾淨分明的 train_df 與 test_df，方便後續特徵工程與模型訓練。

```python
from google.colab import drive
drive.mount('/content/drive')
```

```python
# ============================================================
# ⭐ STEP 0：Google Drive 路徑設定
# ============================================================
base_path = "/content/drive/MyDrive/"

import pandas as pd
import json

# ============================================================
# ⭐ STEP 1：讀取 final_posts.json
# ============================================================

with open(base_path + "final_posts.json", "r", encoding="utf-8") as f:
    posts = json.load(f)

# ⭐ 修正數據：展开 root/_source/post
data_list = []
for item in posts:
    post_info = item.get("root", {}).get("_source", {}).get("post", {})

    post_id = post_info.get("post_id")
    post_text = post_info.get("text")

    data_list.append({
        "id": post_id,
        "text": post_text
    })

# 建立 DataFrame
posts_df = pd.DataFrame(data_list)


# ============================================================
# ⭐ STEP 2：讀取 CSV 資料
# ============================================================

split_df   = pd.read_csv(base_path + "data_identification.csv")
emotion_df = pd.read_csv(base_path + "emotion.csv")


# ============================================================
# ⭐ STEP 3：依照 split (train/test) 合併 Label
# ============================================================

merged = posts_df.merge(split_df, on="id", how="left")

train_df = merged[merged["split"] == "train"].merge(emotion_df, on="id", how="left")
test_df  = merged[merged["split"] == "test"]

train_df = train_df[["id", "text", "emotion"]].reset_index(drop=True)
test_df  = test_df [["id", "text"]].reset_index(drop=True)

print("訓練集 (train_df) 前幾行:")
print(train_df.head())

print("\n測試集 (test_df) 前幾行:")
print(test_df.head())
```

## 2. Feature Engineering Steps

在特徵工程階段，我主要將前處理後的文字內容轉換成模型可使用的數值特徵。首先，我對每則貼文的文字進一步做了簡單清洗，包括統一轉成小寫、移除標點符號與非文字字元，以降低噪訊並提高後續向量化的品質。

接著，我將情緒標籤轉成模型可以辨識的數字 ID，建立文字與標籤之間的對應關係，方便後續訓練分類器。在文字特徵方面，我抽取了清洗後的文字欄位 (clean_text)，作為 TF-IDF 或其他模型向量化的主要輸入來源。這些步驟的目的都是將原始文字轉換成更規範、可量化的格式，使模型能更有效地學習不同情緒之間的語意差異。

```python
#
!pip install emoji

## Add the code related to the feature engineering steps in cells inside this section
import pandas as pd
import numpy as np
from sklearn.feature_extraction.text import TfidfVectorizer
from scipy.sparse import hstack
import scipy.sparse as sp
import emoji
import re

#
def preprocess_text(text):
    if pd.isna(text): # 處理可能的 NaN 值
        return ""
    text = str(text).lower()

    text = re.sub(r'[^\w\s]', '', text)
    return text


train_df["clean_text"] = train_df["text"].apply(preprocess_text)
test_df["clean_text"] = test_df["text"].apply(preprocess_text)

#

labels = sorted(train_df["emotion"].unique())
label2id = {label: i for i, label in enumerate(labels)}
id2label = {i: label for label, i in label2id.items()}
train_df["label"] = train_df["emotion"].map(label2id).astype(int)

#
X_text_train = train_df["clean_text"]
X_text_test  = test_df["clean_text"]

#
```

## 3. Model Implementation Steps

在模型實作階段，我採用了 Transformers 的 RoBERTa-base 作為主要的情緒分類模型。選擇此模型的原因是 RoBERTa 在英文自然語言處理任務上具有穩定且優異的表現，特別是對於推文這類短文本，能夠有效捕捉情感語氣、詞彙細微差異與上下文資訊。

首先，我將整理好的訓練資料進行分割，保留約 10% 作為驗證集，並使用分層抽樣確保各情緒類別比例一致。這樣可以避免某些類別過少導致模型偏向特定情緒，提升模型泛化能力。

接著載入 RoBERTa 模型及其對應的 tokenizer，並將文字編碼成模型可處理的向量形式，包括 input ids、attention mask 等。為了讓模型能辨識不同的情緒，我也依照資料的標籤數量設定模型的輸出分類數。資料編碼完成後，我將訓練集、驗證集與測試集分別包裝成 PyTorch Dataset，確保 Trainer 能正確讀取。

在訓練設定方面，我使用了小批量訓練（batch size = 8）、適合 transformers 模型的較小學習率，以及 4 個訓練週期。為了提升穩定度，也加入 warmup、weight decay 以及 fp16 訓練以加速訓練與降低 GPU 負擔。評估指標則使用 macro-F1，因為情緒分類屬於多類別問題，且資料不平衡，此指標能更公平地衡量各類別的表現。

最後，我使用 HuggingFace 的 Trainer 來負責整個訓練流程，包括模型訓練、驗證、儲存最佳模型等功能。當 Trainer 初始化完成後，只需執行訓練指令即可開始模型學習。整體流程能確保模型穩定、可重現，也方便後續在測試集上進行預測並提交到 Kaggle。

```python
### Add the code related to the model implementation steps in cells inside this section

import pandas as pd
import numpy as np
import torch
import sys

from sklearn.model_selection import train_test_split
from sklearn.metrics import f1_score

from transformers import (
    AutoTokenizer,
    AutoModelForSequenceClassification,
    TrainingArguments,
    Trainer
)

print("Python Kernel:", sys.executable)

# ----------------------------------------------------
# Step A: Split
# ----------------------------------------------------
train_texts, val_texts, train_labels, val_labels = train_test_split(
    train_df["text"].tolist(),
    train_df["label"].tolist(),
    test_size=0.1,
    random_state=42,
    stratify=train_df["label"].tolist()
)

def compute_metrics(p):
    preds = np.argmax(p.predictions, axis=1)
    macro_f1 = f1_score(p.label_ids, preds, average="macro")
    return {"macro_f1": macro_f1}

# ----------------------------------------------------
# Step B: Load Roberta-base (最穩定)
# ----------------------------------------------------

model_name = "roberta-base"

tokenizer = AutoTokenizer.from_pretrained(
    model_name,
    use_fast=True
)

num_labels = len(train_df["emotion"].unique())
model = AutoModelForSequenceClassification.from_pretrained(
    model_name,
    num_labels=num_labels
)

max_len = 256

train_enc = tokenizer(train_texts, truncation=True, padding=True, max_length=max_len)
val_enc = tokenizer(val_texts, truncation=True, padding=True, max_length=max_len)
test_enc = tokenizer(test_df["text"].tolist(), truncation=True, padding=True, max_length=max_len)

# ----------------------------------------------------
# Step C: Dataset
# ----------------------------------------------------

class EmotionDataset(torch.utils.data.Dataset):
    def __init__(self, encodings, labels=None):
        self.encodings = encodings
        self.labels = labels

    def __getitem__(self, idx):
        item = {k: torch.tensor(v[idx]) for k, v in self.encodings.items()}
        if self.labels is not None:
            item["labels"] = torch.tensor(self.labels[idx])
        return item

    def __len__(self):
        return len(self.encodings["input_ids"])

train_ds = EmotionDataset(train_enc, train_labels)
val_ds = EmotionDataset(val_enc, val_labels)
test_ds = EmotionDataset(test_enc)

# ----------------------------------------------------
# Step D: Training Arguments
# ----------------------------------------------------

training_args = TrainingArguments(
    output_dir="roberta_results",
    eval_strategy="epoch",
    save_strategy="epoch",
    per_device_train_batch_size=8,
    per_device_eval_batch_size=8,
    num_train_epochs=4,
    learning_rate=2.2e-5,
    warmup_ratio=0.989,
    weight_decay=0.01,
    logging_steps=50,
    logging_strategy="steps",
    load_best_model_at_end=True,
    metric_for_best_model="macro_f1",
    fp16=True,
    report_to="none"
)

# ----------------------------------------------------
# Step E: Trainer
# ----------------------------------------------------

trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=train_ds,
    eval_dataset=val_ds,
    tokenizer=tokenizer,
    compute_metrics=compute_metrics
)

print("初始化成功，輸入 trainer.train() 開始訓練")
```

```python
trainer.train()
```

```python
# ===========================
# ⭐ Step 2. 預測 test_ds
# ===========================
preds = trainer.predict(test_ds)
logits = preds.predictions
final_preds = logits.argmax(axis=1)


# ===========================
# ⭐ Step 3. label id → label 名稱
# ===========================
# 假設你有這個：
labels = sorted(train_df["emotion"].unique())
label2id = {label: i for i, label in enumerate(labels)}
id2label = {v: k for k, v in label2id.items()}

test_df["emotion"] = [id2label[p] for p in final_preds]


# ===========================
# ⭐ Step 4. 生成 submission.csv
# ===========================
submission = test_df[["id", "emotion"]]
submission.to_csv("submission.csv", index=False)

print("✔ 已產生 submission.csv，可以下載了！")
```

### 4.1 我嘗試過的不同方法
在參加 Kaggle 競賽的過程中，我針對前處理、特徵工程與模型選擇做了多次嘗試，以找出最適合本次任務的組合。首先，在文本特徵方面，我同時測試過 TF-IDF + 傳統機器學習模型（如 Logistic Regression、SVM）與 預訓練語言模型（如 BERT、DistilBERT、RoBERTa）。傳統模型在資料較小時訓練速度快、表現穩定，但在推文這種語意強烈的文本上，效果明顯不如 Transformer 類模型，因此最終改用 RoBERTa-base。

在 RoBERTa 的訓練過程中，我也嘗試過不同的參數組合，例如：調整最大序列長度（128、256）、調整學習率（2e-5、3e-5）、以及不同的 batch size。也測試過是否加入更多清理規則，例如移除 emoji、保留 emoji、是否保留特殊符號等。實驗發現過度清理反而會讓情緒特徵流失，因此最後採用較簡潔的清理方法，只進行必要的標點與大小寫統一處理。

此外，我也嘗試加入 warmup 與 weight decay，並比較不同 epoch 數量對結果的影響，最終選擇在效能與訓練時間之間取得平衡的參數設定。

### 4.2 我得到的心得

在這次情緒分類任務中，我最大的體會是： 傳統 TF-IDF 雖然直覺，但對短文本的語意捕捉有限；相對地，Transformer 模型因具備 contextual encoding，對具有情緒語氣的句子表現明顯更佳。

我也觀察到，資料清理不宜過度。例如推文中的 emoji 雖然看似雜訊，但實際上往往對情緒判斷具有指標性作用；同樣地，某些特殊符號或語助詞也可能代表特定情緒，因此只有基本的清洗反而能保留更多資訊。

最後，在模型訓練上我發現，macro-F1 比單純 accuracy 更能反映模型的真實表現，尤其是遇到情緒類別不平衡的情況時。透過使用這個指標，我更清楚自己的模型是否真的能處理每一種情緒，而不只是偏好常見的類別。

整體而言，本次競賽讓我更理解 NLP 模型如何處理短文本情緒分類，也更熟悉 Transformer 模型的調參方向，對未來處理類似文本任務有很大的幫助。
